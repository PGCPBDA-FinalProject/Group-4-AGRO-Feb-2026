```python
df = spark.read \
    .option("header","true") \
    .option("inferSchema","true") \
    .csv("/Volumes/project/weather/weat/*.csv")
```

```python
df.select("_metadata.file_name").distinct().show(7,False)
```

```python
from pyspark.sql.functions import col

df = df.withColumn(
    "city",
    col("_metadata.file_name")
)
```

```python
df.select("city").distinct().show()
```

```python
from pyspark.sql.functions import avg

city_temp = df.groupBy("city").agg(
    avg("temperature_2m").alias("Avg_Temperature")
).orderBy("Avg_Temperature", ascending=False)

display(city_temp)
```

```python

city_rain = df.groupBy("city").agg(
    avg("rain").alias("Avg_Rainfall")
).orderBy("Avg_Rainfall", ascending=False)

display(city_rain)
```

```python
city_humidity = df.groupBy("city").agg(
    avg("relative_humidity_2m").alias("Avg_Humidity")
).orderBy("Avg_Humidity", ascending=False)

display(city_humidity)
```

```python
city_wind = df.groupBy("city").agg(
    avg("wind_speed_10m").alias("Avg_WindSpeed")
).orderBy("Avg_WindSpeed", ascending=False)

display(city_wind)
```

```python
city_wind = df.groupBy("city").agg(
    avg("wind_speed_10m").alias("Avg_WindSpeed")
).orderBy("Avg_WindSpeed", ascending=False)

display(city_wind)
```

```python
city_temp.orderBy("Avg_Temperature", ascending=False).show(10, False)
```

```python
from pyspark.sql.functions import count

city_records = df.groupBy("city").agg(
    count("*").alias("Record_Count")
).orderBy("Record_Count")

display(city_records)
```

```python
city_temp.orderBy("Avg_Temperature").show(10, False)
```

```python
from pyspark.sql.functions import avg

rain_city = df.groupBy("city").agg(
    avg("rain").alias("Average_Rainfall")
)

display(rain_city.orderBy("Average_Rainfall", ascending=False).limit(10))
```
