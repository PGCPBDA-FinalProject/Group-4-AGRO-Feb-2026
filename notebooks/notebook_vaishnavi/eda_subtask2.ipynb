```python
df = spark.read \
    .option("header","true") \
    .option("inferSchema","true") \
    .csv("/Volumes/project/weather/weat/*.csv")
```

```python
df.printSchema()
```

```python
display(df)
```

```python
df.select("_metadata.file_name").distinct().show(7,False)
```

```python
from pyspark.sql.functions import col

df = df.withColumn(
    "city",
    col("_metadata.file_name")
)
```

```python
df.select("city").distinct().show()
```

```python
print("Rows:", df.count())
print("Columns:", len(df.columns))
```

```python
df.describe().show()
```

```python
from pyspark.sql.functions import *

df.select([
    count(when(col(c).isNull(), c)).alias(c)
    for c in df.columns
]).show()
```

```python
df.select("city").distinct().count()
```

```python
from pyspark.sql.functions import avg

df.groupBy("city") \
  .agg(avg("temperature_2m").alias("avg_temp")) \
  .orderBy(desc("avg_temp")) \
  .show()
```

```python
df.groupBy("city") \
  .agg(sum("precipitation").alias("total_rain")) \
  .orderBy(desc("total_rain")) \
  .show()
```

```python
df.groupBy("city") \
  .agg(avg("wind_speed_10m").alias("avg_wind")) \
  .orderBy(desc("avg_wind")) \
  .show()
```

```python
df.groupBy("city") \
  .agg(avg("wind_speed_100m").alias("avg_wind")) \
  .orderBy(desc("avg_wind")) \
  .show()
```

```python
from pyspark.sql.functions import avg, stddev, min, max

df.select(
    avg("temperature_2m").alias("Avg Temp"),
    stddev("temperature_2m").alias("Std Dev"),
    min("temperature_2m").alias("Min Temp"),
    max("temperature_2m").alias("Max Temp")
).show()
```

```python
q1,q3 = df.approxQuantile("temperature_2m",[0.25,0.75],0)

iqr = q3-q1

lower = q1-1.5*iqr
upper = q3+1.5*iqr

print(lower,upper)
```

```python
df.filter(
    (df.temperature_2m<lower)|
    (df.temperature_2m>upper)
).count()
```

```python
df.filter(df.temperature_2m < -20).count()
```
