```python
import os
import re
import pandas as pd

base_path = "/kaggle/input/datasets/mukeshdevrath007/indian-5000-cities-weather-data"

folders = [
    "w_d_1",
    "w_d_2",
    "w_d_3/w_d_3"
]

records = []

for folder in folders:

    folder_path = os.path.join(base_path, folder)

    if not os.path.exists(folder_path):
        continue

    for file in os.listdir(folder_path):

        if file.endswith(".csv"):

            original_name = os.path.splitext(file)[0]

            # Remove suffix (_2, _3...)
            base_city = re.sub(r'_\d+$', '', original_name)

            records.append([
                base_city,
                original_name,
                folder,
                file
            ])

df = pd.DataFrame(
    records,
    columns=[
        "Base_City",
        "Original_Name",
        "Folder",
        "CSV_File"
    ]
)

# Keep only duplicate base cities
duplicates = (
    df.groupby("Base_City")
      .filter(lambda x: len(x) > 1)
      .sort_values(["Base_City","Folder","Original_Name"])
)

print("Duplicate Base Cities :", duplicates["Base_City"].nunique())
print("Total Files :", len(duplicates))

display(duplicates)

# Export
duplicates.to_csv(
    "/kaggle/working/duplicate_cities_all_folders.csv",
    index=False
)

print("Saved Successfully!")
```

```python
import os
import re
import pandas as pd

# Base path of the Kaggle dataset
base_path = "/kaggle/input/datasets/mukeshdevrath007/indian-5000-cities-weather-data"

# Folder names
folders = [
    "w_d_1",
    "w_d_2",
    "w_d_3/w_d_3"
]

records = []

# Read all files from all folders
for folder in folders:

    folder_path = os.path.join(base_path, folder)

    if not os.path.exists(folder_path):
        continue

    for file in os.listdir(folder_path):

        if file.endswith(".csv"):

            original_name = os.path.splitext(file)[0]

            # Remove suffix (_2, _3, ...)
            base_city = re.sub(r'_\d+$', '', original_name)

            records.append({
                "Base_City": base_city,
                "Original_Name": original_name,
                "Folder": folder,
                "CSV_File": file
            })

# Create DataFrame
df = pd.DataFrame(records)

# Keep only duplicate Base_City values
duplicates = (
    df.groupby("Base_City")
      .filter(lambda x: len(x) > 1)
      .sort_values(["Base_City", "Folder", "Original_Name"])
      .reset_index(drop=True)
)

# Add count of each duplicate city
duplicates["Count"] = duplicates.groupby("Base_City")["Base_City"].transform("count")

# Rearrange columns
duplicates = duplicates[
    [
        "Base_City",
        "Count",
        "Original_Name",
        "Folder",
        "CSV_File"
    ]
]

# Print summary
print("="*60)
print("Total Duplicate Base Cities :", duplicates["Base_City"].nunique())
print("Total Duplicate Files       :", len(duplicates))
print("="*60)

# Display result
display(duplicates)

# Save CSV
output_path = "/kaggle/working/duplicate_cities_all_folders.csv"

duplicates.to_csv(output_path, index=False)

print("\nCSV saved successfully!")
print("Location:", output_path)
```
