```python
df = spark.read.csv("/Volumes/project/weather/weat/crop data/weather (2).csv", header=True, inferSchema=True)
```

```python
from pyspark.sql.functions import collect_list, countDistinct

cities_by_state = (
    df.groupBy("state")
      .agg(
          collect_list("city").alias("Cities"),
          countDistinct("city").alias("City_Count")
      )
      .orderBy("state")
)

display(cities_by_state)
```

```python
from pyspark.sql.functions import countDistinct

total_city_count = df.agg(countDistinct("city").alias("Total_City_Count"))
display(total_city_count)
```
